```python
score = int(input('请输入一个分数：'))

if score in range(0, 60):
    print('等级为：F')
elif score in range(60, 70):
    print('等级为：D')
elif score in range(70, 80):
    print('等级为：C')
elif score in range(80, 90):
        print('等级为：B')
elif score in range(90, 101):
        print('等级为：A')

else:
    print('输入的数字不可小于 0 或大于 100 ！！！')
```
