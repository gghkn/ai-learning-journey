```python
# for循环方法一：
for i in range(2, 101, 2):
    print(i)

```

```python
# for循环方法二：
number = [x*2 for x in range(51) if x != 0]
for i in number:
    print(i)
```

```python
# while循环方法一：
i = 1
while True:
    print(i * 2)
    i += 1
    if i == 51:
        break
    else:
        continue

# print(i)
```

```python
# while循环方法二：
i = 1
i = i * 2
while i < 101:
    print(i)
    i += 2
```
